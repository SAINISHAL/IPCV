# Eigenfaces Face Recognition: Mathematical Exploration
This notebook demonstrates the mathematical concepts of **Principal Component Analysis (PCA)** and **Eigenfaces** (Turk and Pentland, 1991) applied to classroom facial feature representation.

## 1. Load Trained Eigenfaces Model
